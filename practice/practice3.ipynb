```python
def is_prime(a):
    lst3 = []
    for i in a :
        if i == 1 :
            continue
        z = True 
        for q in range(2,i) :
            if i % q == 0 :
                z = False
        if z == True :
            lst3.append(i)

    return lst3
```

```python
x = is_prime(lst1)
print(x)
```

```python
def factorial (input1) :
    if input1 < 0 :
        return 'error'
    elif input1 == 0 :
        return 1 
    else :
        return (input1 * factorial(input1-1))
```

```python
while True : 
    input1 = input('please enter a number:')
    if input1 == 'done' : break
    factorial_all = factorial(int(input1))
    print(factorial_all)
```
