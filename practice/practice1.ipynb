```python
def add_number (input_1 , input_2) :
    input_3 = []
    for i in range(input_1 , input_2) :
        if (i == 0) or (i == 1) :
            continue 
        w = True
        for j in range(2 , i) :
            if i % j == 0 :
                w = False
        if w == True :
            input_3.append(i)
    return input_3
```

```python
input_1 = int(input('palse enter a number :'))
input_2 = int(input('palse enter a number :'))
e = add_number(input_1 , input_2)
print(e)
```
