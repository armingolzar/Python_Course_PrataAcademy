```python
records = [
    ('Armin', [18, 20, 19.5, 17]),
    ('Arad', [12, 10, 14.3, 15]),
    ('Mohammad', [7, 20, 11.6, 17]),
    ('Erfan', [17.8, 16.5, 19, 10]),
    ('Arta', [17, 19, 20, 19])
]
```

```python
dict1 = {}
for key , val in records :
    avg = sum(val) // len(val)
    max_lst = max(val)
    min_lst = min(val)
    lst2 = avg , max_lst , min_lst
    dict1[key] = lst2
    
print(dict1)
```

```python
sentences = corpus.split(',')
clean_sentences = []
for i in sentences:
    i = i.strip()
    clean_sentences.append(i)
clean_sentences.remove('')
print(clean_sentences)

dict_letters = {}
for index, sentence in enumerate(clean_sentences):
    for index2, letter in enumerate(sentence):
        if letter not in (dict_letters):
            dict_letters[letter] = [1, (index, index2)]
        else:
            dict_letters[letter][0] += 1
            dict_letters[letter].append((index, index2))

for output in dict_letters:
    print(output, dict_letters[output])
    print('\n')
        
```
