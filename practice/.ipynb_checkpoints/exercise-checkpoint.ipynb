```python
#calculate the valume of a rectangle
length = float(input("please give me a length\n"))
width = float(input("please give me a width\n"))
height = float(input("please give me a height\n"))
valume = length * width * height
print("valume of the rectangle is :" ,valume)
```

```python
#calculate the valume of sphare
radiuse = float(input("please giv me a radiuse\n"))
pi = 3.14
valume = 4/3*pi*radiuse**3
print("valume of sphare is :",valume)
```
