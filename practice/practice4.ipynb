```python
def prime_nubers ( start , end ) : 
    lst_1 = []
    start = int(start)
    end = int(end)
    for i in range(start , end) :
        if i == 1 : 
            continue 
        x = True 
        for r in range(2 , i) :
            if i % r == 0 :
                x = False 
                break
                
        if x == True :
            lst_1.append(i)

    return lst_1
```

```python
input1 = input('please enter a number :')
input2 = input('please enter a number :')
prime_lst = prime_nubers (input1 , input2) 
print(prime_lst)
```

```python
def make_list ( lst1 ) :
    lst_odd = []
    lst_even = []
    lst_prime = []
    for i in lst1 :
        if i % 2 == 0 :
            lst_even.append(i)
        else :
            lst_odd.append(i)
        f = True
        if i == 1 :
            continue
        for r in range(2 , i) :
            if i % r == 0 :
                f = False
                break
        if f == True :
            lst_prime.append(i)
    return lst_odd , lst_even , lst_prime
```

```python
lst_asghar = [1,2,97,33,48,50,121,17,64,59]
lst_odd,lst_even,lst_prime = make_list (lst_asghar)
print(lst_odd)
print(lst_even)
print(lst_prime)
```

```python
def make_list (lst1 , lst2) :
    lst3 = [] 
    for i in lst1 :
        for q in lst2 :
            if i == q : 
                lst3.append(q)

    return lst3
```

```python
lst1 = [ 2,8,12,25,100]
lst2 = [3,6,12,20,100]
y = make_list (lst1 , lst2)
print(y)
```
