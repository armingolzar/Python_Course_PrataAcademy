```python
def type_counter(test_list) :
    intiger = 0
    string = 0
    tuplee = 0
    dictionary = 0
    none_type = 0 
    boolian = 0 
    floatt = 0
    final_dict = {}
    for i in test_list:
        if type(i) == int :
            if 'int' in final_dict:
                final_dict['int'] += 1
            else:
                final_dict['int'] = 1
        elif type(i) == str :
            if 'str' in final_dict:
                final_dict['str'] += 1
            else:
                final_dict['str'] = 1
        elif type(i) == tuple :
            if 'tup' in final_dict:
                final_dict['tup'] += 1
            else:
                final_dict['tup'] = 1
        elif type(i) == dict :
            if 'dic' in final_dict:
                final_dict['dic'] += 1
            else:
                final_dict['dic'] = 1
        elif i == None :
            if 'non' in final_dict:
                final_dict['non'] += 1
            else:
                final_dict['non'] = 1
        elif type(i) == bool :
            if 'bol' in final_dict:
                final_dict['bol'] += 1
            else:
                final_dict['bol'] = 1
        elif type(i) == float:
            if 'flo' in final_dict:
                final_dict['flo'] += 1
            else:
                final_dict['flo'] = 1

    return final_dict
```
