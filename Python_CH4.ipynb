## The while statement

```python
n=5
while n > 0:
    print(n)
    n = n-1 
print('Done!')
```

### How it works
 1. Evaluate the condition, yielding True or False.
 2. If the condition is false, exit the while statement and continue execution at the next statement.
 3. If the condition is true, execute the body and then go back to step 1.

## Infinite Loops

```python
n = 10 
while n <= 10:
    print(n, end=' ')
    n = n - 1
    
print('Done!')
```

```python
n = 10
try:
    while True:
        print(n, end = ' ')
        n = n + 1
except:
    print('\n\nOver!')
```

## Finishing Iterations with break & continue

```python
## use break in infinite loops
while True:
    line = input('>>> ') 
    if line == 'done':
        break
    print(line)
print('Done!')
```

Sometimes you are in an iteration of a loop and want to finish the current iteration and immediately jump to the next iteration. In that case you can use the ***continue*** statement to skip to the next iteration without finishing the body of the loop for the current iteration.


```python
while True:
    line = input('>>> ') 
    if line == 'hi':
        # continue
        print('sallaaaammmmm')
    if line == 'hi':  #### why we use if and do not use elif here?
        break
    print(line)


print('Done!')
```

## Quiz 
- Calculate n! using loops for a given n.

- Write a program such that the output be "Aa, Bb, Cc, ..., Zz".

```python
# calculate n!:
while True:

    try:
        n = input('Enter a non-negative integer: ')
        if n == 'done':
            break
        n = int(n)    
        if n >= 0:
            factorial = 1
            for i in range(1,n+1):
                factorial = factorial * i

            print(factorial)
        else:
            print('n should be non-negative')

    except:
        print('Type Error!')
    
```
